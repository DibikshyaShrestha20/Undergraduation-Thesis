```python
import matplotlib.pyplot as plt
import numpy as np


conditions = ['COPD', 'Asthma', 'Lung Cancer']
pre_covid = np.array([112, 32, 6])
during_covid = np.array([22, 7, 2])
after_covid = np.array([116, 39, 8])

#  Bar Chart
fig, ax1 = plt.subplots(figsize=(10, 6))

bar_width = 0.35
x = np.arange(len(conditions))

rects1 = ax1.bar(x, pre_covid, bar_width, label='Pre-COVID', color='skyblue')
rects2 = ax1.bar(x, during_covid, bar_width, bottom=pre_covid, label='During COVID', color='lightcoral')
rects3 = ax1.bar(x, after_covid, bar_width, bottom=pre_covid + during_covid, label='After COVID', color='lightgreen')

# Customization
ax1.set_ylabel('Number of Cases')
ax1.set_xlabel('Medical Condition')
ax1.set_title('Number of Cases of Respiratory Conditions Before, During, and After COVID-19 \n T.U. Teaching Hospital')
ax1.set_xticks(x)
ax1.set_xticklabels(conditions)
ax1.legend()

ax1.bar_label(rects1, label_type='center')
ax1.bar_label(rects2, label_type='center')
ax1.bar_label(rects3, label_type='center')

# Pie Chart
total_pre = np.sum(pre_covid)
total_during = np.sum(during_covid)
total_after = np.sum(after_covid)
total_cases = [total_pre, total_during, total_after]
labels_pie = ['Pre-COVID', 'During COVID', 'After COVID']
colors_pie = ['skyblue', 'lightcoral', 'lightgreen']

# Subplot
ax2 = fig.add_axes([0.65, 0.65, 0.25, 0.25]) # [left, bottom, width, height]
ax2.pie(total_cases, labels=labels_pie, autopct='%1.1f%%', startangle=140, colors=colors_pie)
ax2.axis('equal')
ax2.set_title('Total Cases by Period')

fig.tight_layout(rect=[0, 0, 0.6, 1])
plt.show()
```

```python
#  save the figure
plt.savefig('respiratory_cases_analysis.png')
```
